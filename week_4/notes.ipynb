```python
# Tuples are immutable at the top level

T = (1, [2, 3], 4)
T[1] = 'changed'
```

```python
# Modifying a mutable element inside a tuple

T[1][0] = 'changed'
T
```

```python
# a single-item tuple
single_item_tuple = (47.3)
type(single_item_tuple)
single_item_tuple[0] = 77.0

# TODO: a point of extra credit for explaining 
# if this is a quirk of python or intentional 
```

```python
# a single-item tuple with a trailing comma
single_item_tuple = (47.3,)
type(single_item_tuple)
```

```python
# named tuples
# named tuples are immutable like regular tuples but allow access to elements by name
from collections import namedtuple

Breakfast = namedtuple('Breakfast', ['guests', 'eggs', 'bagels'])
p = Breakfast(guests=11, eggs=22, bagels=33)
p
```

```python
# list comprehensions but with tuples
# this concept applies to lists AND tuples
T = (5, 4, 3, 2, 1)
T_squared = tuple(x**2 for x in T)
T_squared
L_squared = [x**2 for x in T]
L_squared
```

```python
# Indexing and slicing 
T = (5, 4, 3, 2, 1)
T[0], T[1:4] # create a tuple with nested objects

nested_tuple = (T[0], T[1:4])
type(nested_tuple)
```

```python
# writing data to a file
myfile = open('myfile.txt', 'w')
myfile.write('Hello, world!\n')
myfile.write('Goodbye, world!\n')
myfile.close()
```

```python
myfile = open('myfile.txt', 'r') # read is the default, so I dont have to include the r

content_full = myfile.read()
content_line = myfile.readline()
myfile.close()
print("content_full:", content_full)
print("content_line:", content_line)
```
