```python
# Example 1: List comprension
L = []
for x in range(10):
    L.append(x+10)
print(L)

# Using list comprehension
L = [x+10 for x in range(10)]
print(L)
```

```python
# Example 2: List comprehension with timing
import time

L = []
start_time = time.time()
for x in range(1000):
    L.append(x**2+10)
# print(L)
end_time = time.time()
print(f"{end_time - start_time} seconds elapsed")

# Using list comprehension
start_time = time.time()
L = [x**2+10 for x in range(1000)]
# print(L)
end_time = time.time()
print(f"{end_time - start_time} seconds elapsed")
```

```python
# Example 3: List of states with inconsistent formatting
states = [' Alabama ', 'Georgia!', 'Georgia', 'georgia',
'FlOrIda', 'south carolina##', 'West virginia?']

states2 = []
#for loop
for s in states: 
    states2.append(s.strip().title().replace('!', '').replace('##', '').replace('?', ''))
print(states2)

# Using list comprehension
states3 = [s.strip().title().replace('!', '').replace('##', '').replace('?', '') for s in states]
print(states3)
```

```python
# example 4: create a pumpkin
def draw_pumpkin():
    """
    Draws a simple pumpkin using ASCII art.
    Returns:
        None    
    """
    print(" _/_")
    print("/~,~\\")
    print("\\\_//")

draw_pumpkin()
```

```python
# example 5 - adding numbers
def add_numbers(a: int | float, b: int | float) -> int | float:
    """
    Adds two numbers together.

    Args:
        a (int or float): The first number.
        b (int or float): The second number.

    Returns:
        int or float: The sum of the two numbers.
    """
    y = a + b
    return y

output = add_numbers(3, 5)
print(output)
```

```python
# Example 6 - adding two lists
list1 = [1, 2, 3]
list2 = [4, 5, 6]

def add_lists(L1: list, L2: list) -> tuple[list, list, list]:
    """
    Adds two lists together.

    Args:
        L1 (list): The first list.
        L2 (list): The second list.

    Returns:
        tuple[list, list, list]: A tuple containing the concatenation of the two lists, followed by the original lists.
    """

    return L1 + L2, L1, L2
result, first, second = add_lists(list1, list2)
print(result)
print(first)
print(second)
```

```python
# Example 7 - Global scope
# global can be relative or absolute
X = 99
def func(Y):
    Z = X+Y # X is global to the function and Y is local to the function
    return Z
print(func(1))
```

```python
# Example 8 - Using global keyword inside a function
#  X = 88 
def func():
    global X # this promotes X to a global variable which is a bad practice
    X = 88

    Y = X+1
    return Y
print(func())
print(f"We have the value of X outside the function:{X}")
type(X)
```
