```python
class Person:
    # overloading by setting age to None by default
    def __init__(self, name, jobs, age=None):
        self.name = name
        self.age = age
        self.jobs = jobs

    def info(self):
        print(f"name: {self.name}, age: {self.age}, jobs: {self.jobs}")
```

```python
rec1 = Person("Bob", ["dev","mgr"], 40.5) #create instance by calling constructor

rec1.info()
```

```python
rec2 = Person("Alice", ["analyst"])
rec2.info()
```

```python
# class 1 inherits 2 and 3
class Class2:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class Class3:
    def __init__(self, z):
        self.z = z

class Class1(Class2, Class3):
    def __init__(self, name):
        self.name = name
        Class2.__init__(self, x=0, y=5)
        Class3.__init__(self, z=10)
```

```python
# create an instance of Class1
instance1 = Class1("bob")
print(instance1.z)
```

```python
# method overloading
from multipledispatch import dispatch

class Person:
    def __init__(self, name):
        self.name = name

    @dispatch(str, int)
    def new(self, jobs, salary):
        self.jobs = jobs
        self.salary = salary

    #overload the method
    @dispatch(str, int, int)
    def new(self, jobs, salary, age): 
        self.age = age
        self.jobs = jobs
        self.salary = salary

    def info(self):
        print(f"name: {self.name}, age: {self.age}, jobs: {self.jobs}")
```

```python
rec2 = Person("Alice")
rec2.new("Engineer", 100000, 30)
rec2.new("Pizza Chef", 50000)

rec2.info()
```
