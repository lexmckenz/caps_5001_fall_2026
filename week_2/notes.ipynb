```python
# Example 1: Explicit Conversion - you specify
# the desired object type explicitly
x = "123"
y = int(x)
y

x = int(5.0)
print(x)
```

```python
# Example 2: Implicit Conversion - Python automatically 
# converts the object type

x = 5 + 5
print(x)
y = 5.0 + 5.0
print(y)
z = 5 + 5.03 # floating point error 
print(z)
```

```python
# Example 3: Equals (==) - checks if 
# two values are equal
x = 5
y = 6
print(x == y)

l = [1,2,3]
l2 = l
print("1. l == l2:")
print(l == l2)

l2 = [2,1,3]
print("2. l == l2:")
print(l == l2)


# Example 4: Not Equals (!=) - checks if 
# two values are not equal
x = 5
y = 10
print("x != y:")
print(x != y)
```

```python
# Example 5: is - 
# checks if two variables point to the same 
# location in memory
x = [1, 2, 3]
y = x
print("x is y:")
print(x is y)

y = [1, 2, 3]
print("x is y:")
print(x is y)
```

```python
# Example 6 
# use the l.append(7)list function to add a new 
# list element of number 7 to the list l 
# check the value of l2 to see if l2 gets updated
# with the new list element automatically
l = [1, 2, 3]
l2 = l
l.append(7)
print(l)
print(l2)
```

```python
# Example 7: id() - checks the memory address of
#  a variable
x = [1, 2, 3]
y = x
print("id(x):")
print(id(x))
print("id(y):")
print(id(y))

y = [1, 2, 3]
print("id(x):")
print(id(x))
print("id(y):")
print(id(y))
```

```python
# Example 8: copy() - creates a shallow copy 
# of a list
l = [1, 2, 3]
l2 = l.copy()
print("l:")
print(l)
print(id(l))
print("l2:")
print(l2)
print(id(l2))

# Modify the original list and check if the 
# copy is affected
l.append(7)
print("After appending 7 to l:")
print("l:")
print(l)
print("l2:")
print(l2)
```

```python
# Example 9: slice a string from a longer string
s = "asdhave"
to_find = s.find("ha")
length = len("ha")
sliced = s[to_find:to_find+length]
print(sliced)
```
