```python
import pandas as pd

df_clothes = pd.DataFrame(data=[["blue", "S", 10, "shirt"],
                                ["pink", "M", 120, "peacoat"],
                                ["red", "M", 2000, "tracksuit"],
                                ["green", "L", 15, "hat"]])

df_clothes.columns = ["color", "size", "price", "type"]

df_clothes
```

```python
# Converting categorical values to numerical
# size is ordinal, so it can be easily encoded

size_mapping = {"S": 1, "M": 2, "L": 3}
df_clothes["size_modified"] = df_clothes["size"].map(size_mapping)
df_clothes
```

```python
# one hot encoding
# this will not work if you have data that is already numerical
df_clothes = pd.get_dummies(df_clothes, columns=["color"])
df_clothes
```

```python
# good for nominal values
#when we want a number for nominal categorical values
# but we recognize that the number itself doesnt mean anything

from sklearn.preprocessing import LabelEncoder
class_labels = LabelEncoder()
df_clothes["type_modified"] = class_labels.fit_transform(df_clothes["type"])
df_clothes
```
