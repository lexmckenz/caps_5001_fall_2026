Example 1: Use a try/except with a multi-line block.

```python
print("Printing some text")
x = 5
try: 
    print("Here is a print statement in the try")
    z = spam + x
    print("Does this line execute")
except NameError:
    print("Caught a NameError exception")
    print("We encountered a name error")
```

Example 2: including the error message from the default handler but putting your own flair on the message

```python
print("Printing some text")
x = 5
try: 
    print("Here is a print statement in the try")
    z = spam + x
    print("Does this line execute")
except NameError as e:
    print("Caught a NameError exception")
    print(f"We encountered a name error: {e}")
```

Example 3 : A try/except where no exception occurs

```python
print("Printing some text")
x = 5
try: 
    print("Here is a print statement in the try")
    z = 7 + x
    print("Does this line execute")
except NameError as e:
    print("Caught a NameError exception")
    print(f"We encountered a name error: {e}")
```

Example 4: Raising an exception yourself with the raise keyword. This triggers an exception through the code and specifies one of the default types.

```python
try:
    raise IndexError
except IndexError:
    print("Caught an IndexError exception")
```

```python
# Example 5: raise with finally
try:
    raise IndexError
except IndexError:
    print("Caught an IndexError exception")
finally:
    print("This is the finally block")
```

```python
# example 6: what happens if theres no exception
try:
    # no exception is raised here
    print("No exception occurred in the try block")
except IndexError:
    print("Caught an IndexError exception")
finally:
    print("This is the finally block")
```

```python
# example 7: using else with try-except
try:
    # raise IndexError
    print("Trying something that won't raise an exception")
except IndexError:
    print("Caught an IndexError exception")
else:
    print("This is the else block, executed because no exception occurred")
```

```python
# example 8: combining try-except-else with multiple exceptions
try:
    # raise IndexError
    # raise ValueError
    print("Trying something that might raise an exception")
except IndexError:
    print("Caught an IndexError exception")
except ValueError:
    print("Caught a ValueError exception")
else:
    print("This is the else block, executed because no exception occurred")
finally:
    print("This is the finally block")
```

```python
#example 9: except without exception type
try:
    z = 1/0
except:
    print("Caught an exception without specifying the exception type")
```

```python
# example 10: Exception exception object
try:
    x = int("not a number")
# except Exception:
#     print("Caught a general exception")
except Exception as e:
    print(f"Caught an exception: {e}")
```

```python
# example 11: repr function
try:
    x = int("not a number")
except Exception as e:
    print(f"Caught an exception: {repr(e)}")
```

# Classes

```python
class a_sample_class:
    #default/basic constructor
    def __init__(self): 
        self.var = 0
    #an instance method for printing
    def print_vars(self):
        print(self.var)

#create instance of class
obj = a_sample_class()

#call instance method to get value for var
obj.print_vars()
```

```python
class a_sample_class:
    class_attr = 5
    #parameterized constructor
    def __init__(self, one, two): 
        self.first = one
        self.second = two

    #an instance method for printing
    def print_vars(self):
        print(f"first: {self.first}, second: {self.second}")

#create instance of class
obj = a_sample_class(17,42)

#call instance method to get values
#for inserted parameters
obj.print_vars()
```
