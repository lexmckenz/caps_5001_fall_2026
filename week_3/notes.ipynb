```python
# 1. how do we know what types were interacting with? 

list_example = [1,2,3]
print(type(list_example))

dict_example = {"eggs":12,"ham":1,"bacon":0}
print(type(dict_example))
```

```python
#2. Lists are mutable, which means their elements
# can be changed in place. a_list can be changed without
# needing to create a new variable to reflect
# the change.
a_list = [1, 2, 3]
print(f"Original value of a list: {a_list}")
a_list[0] = 10
print(f"Modified value of a list: {a_list}")
```

```python
# 3. iterating through a list
x = [10,20,30,40,50]

#how long is the list? 
list_len = len(x)
print(list_len)

for i in range(list_len):
    print(f"x[{i}] = {x[i]}")

```

```python
# 4. nested lists
nested_list = [[1, 2, 3, 4], 
               [4, 5, "cat"], 
               [7, 8, 9]]
print(nested_list)

```

```python
# ROWS in the list
rows = len(nested_list)
print(rows)
#COLUMNS in a list
cols = len(nested_list[0])
print(cols)
```

```python
for r in range(rows):
    for c in range(cols):
        print(f"nested_list[{r}][{c}] = {nested_list[r][c]}")
        # this error occurred 
        # EC: 2 pts of extra credit to fix the output so it 
        # doesnt error
```

```python
# 5. slicing lists

x = [10, 20, 30, 40, 50]

# use : at the beginning of a slice
print(x[:2]) # Slice from index 0 to 2 (excluding 2)

# use : at the end of a slice
print(x[2:]) # Slice from index 2 to the end
```
